```python
import re
import string
def clean_text(text):
  text = text.lower()
  text = re.sub(r'\s', ' ',text)
  text = text.translate(str.maketrans('','',string.punctuation))
  return text.strip()
```

```python
def classify_document(text):
  text = text.lower()
  if any(word in text for word in ["invoice","invoice number","billing","subtotal","tax"]):
    return"Invoice"
  elif any(word in text for word in ["resume","curriculum vitae","experience","education","skills"]):
    return"Resume"
  elif any(word in text for word in ["certificate","certification","certificate ID"]):
    return"Certificate"
  elif any(word in text for word in ["passport","date of birth","nationality"]):
    return"ID Document"
  else :
    return "Other"
```

```python
import re
```

```python
def extract_invoice_info(text):

    info = {
        "invoice_number": None,
        "date": None,
        "total": None,
        "tax": None
    }

    # Invoice number
    invoice_match = re.search(
        r'(?:invoice\s*(?:number|no|#)?\s*[:\-]?\s*)([A-Z0-9\-]+)',
        text,
        re.IGNORECASE
    )

    if invoice_match:
        info["invoice_number"] = invoice_match.group(1)

    # Date
    date_match = re.search(
        r'(\d{1,2}[\/\-]\d{1,2}[\/\-]\d{2,4})',
        text
    )

    if date_match:
        info["date"] = date_match.group(1)

    # Total
    total_match = re.search(
        r'(?:total\s*(?:amount)?|grand\s*total)\s*[:\-₹Rs.\s]*([\d,]+(?:\.\d+)?)',
        text,
        re.IGNORECASE
    )

    if total_match:
        info["total"] = total_match.group(1)

    # Tax
    tax_match = re.search(
        r'tax\s*[:\-₹Rs.\s]*([\d,]+(?:\.\d+)?)',
        text,
        re.IGNORECASE
    )

    if tax_match:
        info["tax"] = tax_match.group(1)

    return info
```

```python
invoice_text ="""
Invoice Number : INV-5001
Date : 17/08/2026

Product : Laptop
Quantity : 1
Price : 55000
Tax : 9900
Total Amount : 64900
Payment Due : 30 Augest 2026
"""
```

```python
invoice_info = extract_invoice_info(invoice_text)
print(invoice_info)
```

```python
import re
```

```python
def extract_resume_info(text):

    info = {
        "email": None,
        "phone": None,
        "skills": [],
        "education": []
    }

    # Email
    email_match = re.search(
        r'[\w\.-]+@[\w\.-]+\.\w+',
        text
    )

    if email_match:
        info["email"] = email_match.group()

    # Phone number
    phone_match = re.search(
        r'(?:\+91[\s-]?)?[6-9]\d{9}',
        text
    )

    if phone_match:
        info["phone"] = phone_match.group()

    # Skills
    skill_list = [
        "python",
        "java",
        "sql",
        "html",
        "css",
        "machine learning",
        "pandas",
        "numpy",
        "git"
    ]

    text_lower = text.lower()

    for skill in skill_list:
        if skill in text_lower:
            info["skills"].append(skill.title())

    # Education
    education_list = [
        "bca",
        "btech",
        "mca",
        "mba",
        "bsc",
        "msc"
    ]

    for degree in education_list:
        if degree in text_lower:
            info["education"].append(degree.upper())

    return info
```

```python

resume_text = """
Aryan Choudhary

Email: aryan@example.com
Phone: 9876543210

Education:
BCA

Skills:
Python
SQL
HTML
CSS
Machine Learning
Git

Experience:
Python Developer Intern
"""
```

```python
resume_info = extract_resume_info(resume_text)

print("📄 Resume Information")
print("----------------------")

for key, value in resume_info.items():
    print(f"{key}: {value}")
```

```python
def extract_certificate_info(text):

    info = {
        "certificate_id": None,
        "date": None,
        "course": None
    }

    # Certificate ID
    id_match = re.search(
        r'certificate\s*(?:id|number|no)\s*[:\-]?\s*([A-Z0-9\-]+)',
        text,
        re.IGNORECASE
    )

    if id_match:
        info["certificate_id"] = id_match.group(1)

    # Date
    date_match = re.search(
        r'\d{1,2}[/-]\d{1,2}[/-]\d{2,4}',
        text
    )

    if date_match:
        info["date"] = date_match.group()

    # Course
    course_match = re.search(
        r'(?:course|program)\s*[:\-]\s*(.+)',
        text,
        re.IGNORECASE
    )

    if course_match:
        info["course"] = course_match.group(1).strip()

    return info
```

```python
certificate_text = """
Certificate of Completion

This certificate is awarded for successfully
completing the training program.

Course: Python and Machine Learning

Certificate ID: PY-2026-101

Date: 17/08/2026
"""
```

```python

certificate_info = extract_certificate_info(certificate_text)

print("🏆 Certificate Information")
print("---------------------------")

for key, value in certificate_info.items():
    print(f"{key}: {value}")
```

```python
def analyze_document(text):

    document_type = model.predict([text])[0]

    print("Document Type:", document_type)
    print("--------------------------")

    if document_type == "Invoice":
        return extract_invoice_info(text)

    elif document_type == "Resume":
        return extract_resume_info(text)

    elif document_type == "Certificate":
        return extract_certificate_info(text)

    else:
        return {
            "message": "No specific information extractor available."
        }
```

```python
result = analyze_document(resume_text)

print("\nExtracted Information:")

for key, value in result.items():
    print(f"{key}: {value}")
```
